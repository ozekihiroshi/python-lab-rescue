## 1.6.1 Repeat over known values with for and range

Write `for temporary_name in values:` and Python runs the indented block once for each item. The loop variable represents the value currently being processed.

```python
weekly_attendance = [28, 31, 32, 34]
for attendance in weekly_attendance:
    print("Attendance this week:", attendance)
print("Processing complete")
```

### range() expresses a count or integer interval

`range(start, stop, step)` includes start and excludes stop. `range(1, 5)` produces 1, 2, 3, 4. State the expected iteration count to expose an off-by-one error.

```python
print(list(range(1, 5)))
for week in range(1, 5):
    print(f"Week {week}")
print(list(range(10, 4, -2)))
```

## 1.6.2 Build totals and conditional counts

An accumulator preserves a value across iterations. Start a total at zero and add the current item each time. Printing the running total makes the update order visible.

```python
weekly_attendance = [28, 31, 32, 34]
total = 0
for attendance in weekly_attendance:
    total += attendance
    print("Running total:", total)
print("Total:", total)
print("Mean:", total / len(weekly_attendance))
```

### Count items that satisfy a condition

A counter also starts at zero and increases by one only when a condition is True. Keep a total and a count in separate variables so each role remains clear.

```python
weekly_attendance = [28, 31, 32, 34]
weeks_at_least_30 = 0
for attendance in weekly_attendance:
    if attendance >= 30:
        weeks_at_least_30 += 1
print(weeks_at_least_30)
```

## 1.6.3 Track an extreme value and its position

Starting a maximum at zero fails when every value is negative. If the data is known to be non-empty, start with its first item. If empty data is possible, handle that case explicitly.

```python
weekly_change = [-4, -2, -7, -3]
if weekly_change:
    largest = weekly_change[0]
    for change in weekly_change[1:]:
        if change > largest:
            largest = change
    print("Largest:", largest)
else:
    print("No data")
```

### enumerate() supplies a position and value together

Use `enumerate()` instead of maintaining a manual counter when both a label and value are required. `start=1` produces reader-facing week numbers from one.

```python
weekly_attendance = [28, 31, 32, 34]
for week, attendance in enumerate(weekly_attendance, start=1):
    print(f"Week {week}: {attendance} learners")
```

## 1.6.4 Control one iteration or the whole loop

`continue` skips the remaining statements for the current item and moves to the next. `break` ends the entire loop. Make the reason for skipping or stopping explicit.

```python
readings = [28, None, 31, -1, 34]
for reading in readings:
    if reading is None:
        continue
    if reading < 0:
        print("Invalid value found; stopping")
        break
    print("Valid value:", reading)
```

## 1.6.5 Repeat until a condition changes with while

Use `while` when a stopping condition rather than a known collection controls repetition. Update the state used by the condition inside the loop; otherwise the loop may never end.

```python
remaining = 3
while remaining > 0:
    print("Remaining:", remaining)
    remaining -= 1
print("Complete")
```

## 1.6.6 Handle zero iterations and empty input

Use for when there is an iterable of values; use while when work continues until a condition changes. A direct for loop usually avoids the boundary and update errors of indexing known data with while.

```python
weekly_attendance = [28, 31, 32, 34]
for attendance in weekly_attendance:
    print(attendance)

# The index-based while version needs explicit boundary and update logic.
index = 0
while index < len(weekly_attendance):
    print(weekly_attendance[index])
    index += 1
```
