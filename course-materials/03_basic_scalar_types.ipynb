## 1.3.1 Distinguish the basic scalar types

`int` represents whole numbers, `float` represents numeric values with a fractional form, `str` represents text, `bool` represents truth values, and `NoneType` represents absence of a value. Use `type()` to inspect the actual value.

```python
values = [34, 82.5, "34", True, None]
for value in values:
    print(repr(value), type(value))
```

## 1.3.2 Calculate with numeric operators and explicit order

Run `+`, `-`, `*`, `/`, `//`, `%`, and `**`. `/` normally returns a `float`; `//` floors the quotient; `%` returns the remainder.

```python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 2)
```

### Make calculation order explicit

Exponentiation, multiplication and division, then addition and subtraction are evaluated in order. Use parentheses when they make the work meaning easier to read.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
books = 53
per_box = 12
print("Full boxes:", books // per_box)
print("Books left:", books % per_box)
```

## 1.3.3 Convert and validate values deliberately

Use `int()`, `float()`, and `str()`. Text that cannot represent the requested value raises `ValueError`. `bool()` tests emptiness rather than understanding the word, so `bool("False")` is still `True`.

```python
learners_text = "36"
learners = int(learners_text)
rate_text = "80.5"
rate = float(rate_text)
print(learners + 4, type(learners))
print(rate, type(rate))
print(str(learners) + " learners")
print(bool("False"), bool(""))
```

### Read a conversion error

The text `"12.5"` cannot be converted directly with `int()`. Convert it to `float` first, then decide whether discarding its fractional part is truly appropriate.

```python
try:
    int("12.5")
except ValueError as error:
    print(type(error).__name__ + ":", error)

value = float("12.5")
print(value, type(value))
```

### Validate numeric results

Division by zero raises `ZeroDivisionError`. Floating-point values are binary approximations, so `0.1 + 0.2` may not display as exact 0.3. Use `round()` for reporting when the chosen precision has a clear reason.

```python
print(0.1 + 0.2)
print(round(0.1 + 0.2, 1))
try:
    print(10 / 0)
except ZeroDivisionError as error:
    print(type(error).__name__ + ":", error)
```
