## 1.2.1 Create and update program state with assignment

Predict total and unused seats before running the cell. Then change only the assignment of `morning` to 20 and rerun.

```python
capacity = 40
morning = 18
afternoon = 12

print("Total:", morning + afternoon)
print("Unused seats:", capacity - (morning + afternoon))
```

### Assignment and comparison

One `=` evaluates the right side and assigns the result to the name on the left. Two signs, `==`, compare values without changing state.

```python
registered = 40
print("Registered:", registered)
print("Equal to 40:", registered == 40)
print("Equal to 35:", registered == 35)
```

### Read the right before updating the left

In `total = total + amount`, current values 12 and 5 are read first. Python makes 17, then updates `total`. `+=` is shorthand for the same state change.

```python
total = 12
amount = 5
print("Before:", total)
total = total + amount
print("After long form:", total)
total += amount
print("After shorthand:", total)
```

### A calculated value does not update itself

Changing only `cancelled` after calculating `delivered` does not change `delivered` automatically. Predict both displayed values before running.

```python
planned = 15
cancelled = 2
delivered = planned - cancelled

cancelled = 4
print("Before recalculation:", delivered)

delivered = planned - cancelled
print("After recalculation:", delivered)
```

## 1.2.2 Choose meaningful names and diagnose NameError

`completed_learners` and `group2` are valid. `2nd_group`, names containing spaces, and the keyword `class` are invalid. Names are case-sensitive. All capitals in `MAX_SEATS` communicate an intention not to change the value, but Python does not enforce it.

```python
MAX_SEATS = 40
print("Initial value:", MAX_SEATS)
MAX_SEATS = 36  # This runs, although it conflicts with the constant convention.
print("After reassignment:", MAX_SEATS)
```
