## 2.1.1 Store and modify ordered values with lists

Create a list with brackets and read values with zero-based indexes, negative indexes, and slices. Grouping values with the same meaning makes repeated processing practical.

```python
attendance = [28, 31, 30, 33]
print(attendance[0])
print(attendance[-1])
print(attendance[1:3])
print(len(attendance))
print(31 in attendance)
```

### Modify list contents

Index assignment, `append()`, `extend()`, `insert()`, `remove()`, and `pop()` change a list. Most mutation methods change the existing list rather than returning a useful new list.

```python
attendance = [28, 31, 30]
attendance[0] = 29
attendance.append(33)
attendance.extend([32, 34])
removed = attendance.pop()
print(attendance)
print("Removed:", removed)
```

### Assignment shares one list; copy() creates another list

`backup = original` gives the same list two names, so a mutation is visible through both. Use `copy()` when an independent shallow copy is required.

```python
original = [28, 31, 30]
shared = original
separate = original.copy()
shared.append(33)
print("original:", original)
print("shared:", shared)
print("separate:", separate)
```

### Iterate over values directly, or use enumerate() for positions

Do not maintain an index manually when it is unnecessary. Use for for values and enumerate when the report also needs a week number.

```python
attendance = [28, 31, 30, 33]
for week, value in enumerate(attendance, start=1):
    print(f"Week {week}: {value} learners")
```

## 2.1.2 Use tuples for fixed ordered values

Create a tuple with parentheses and read it by index or iteration. Use one for a fixed group such as report year and month when its elements should not be replaced.

```python
report_period = (2026, 8)
year, month = report_period
print(year, month)
# report_period[0] = 2027  # TypeError: tuples are immutable
```

## 2.1.3 Represent named fields with dictionaries

A dictionary stores `key: value` pairs. Read by a meaningful key such as `centre["registered"]`, not a numeric position. A key cannot occur twice in the same dictionary.

```python
centre = {
    "name": "North Learning Centre",
    "district": "North",
    "registered": 40,
    "completed": 32,
}
print(centre["name"])
print(centre["completed"] / centre["registered"] * 100)
```

### Decide how a missing key should be handled

Reading a missing key with brackets raises `KeyError`. For an optional field, `get()` can supply a default. Do not hide a missing required field with get; distinguish required and optional operational data.

```python
centre = {"name": "North", "registered": 40}
print(centre.get("phone", "not recorded"))
print("completed" in centre)
centre["completed"] = 32
centre["registered"] = 42
print(centre)
```

### items() iterates over keys and values together

A direct for loop over a dictionary yields keys. Use `items()` for both key and value, `keys()` for keys, and `values()` for values.

```python
centre = {"registered": 40, "attended": 34, "completed": 32}
for field, value in centre.items():
    print(field, value)
```

## 2.1.4 Represent several records with a list of dictionaries

Treat one dictionary as one record and each key as a field name. This structure is the bridge to later CSV tables and DataFrames.

```python
centres = [
    {"name": "North", "district": "A", "registered": 40, "completed": 32},
    {"name": "South", "district": "B", "registered": 35, "completed": 24},
]
for centre in centres:
    rate = centre["completed"] / centre["registered"] * 100
    print(f"{centre['name']}: {rate:.1f}%")
```

## 2.1.5 Represent unique membership with sets

A set focuses on whether a value is present, not its order or duplicates. Union `|`, intersection `&`, and difference `-` compare categories. Do not depend on display order.

```python
offered = {"Python", "Data", "Office"}
requested = {"Python", "Web", "Data"}
print("Common:", offered & requested)
print("Not offered:", requested - offered)
print("All:", offered | requested)
```

### Choose the structure from the purpose

Use a list for ordered mutable values, a tuple for a fixed group, a dictionary for named fields, and a set for unique membership. Use a list of dictionaries for several records.

## 2.1.6 Search, add, update, and remove records in an equipment register

Represent three assets as a list of dictionaries containing `asset_id`, `name`, and `available`. Find one by ID and use `None` when absent. Add non-duplicate `A004`, mark `A002` available, and remove `A003`. Display final ID order and count, and compare with a shallow copy containing only the original three records.


```python
books = [
    {"id": "B001", "title": "Python Basics", "read": False},
    {"id": "B002", "title": "Working with Data", "read": True},
    {"id": "B003", "title": "Useful Functions", "read": False},
]

target_id = "B002"
found = None
for book in books:
    if book["id"] == target_id:
        found = book
        break

print(found)
```

### Add, read, update, and remove records in one collection

Build an ID set before adding so duplicates can be rejected. The dictionary returned by the search is the stored dictionary, so changing one field changes that record in the list. To remove, use `enumerate()` to obtain both position and record, then `pop()` that position. A first-to-last search preserves the order of remaining records. These four operations are often called CRUD: Create, Read, Update, and Delete.

```python
existing_ids = set()
for book in books:
    existing_ids.add(book["id"])

new_book = {"id": "B004", "title": "Files in Practice", "read": False}
if new_book["id"] not in existing_ids:
    books.append(new_book)

found["read"] = False

remove_index = None
for index, book in enumerate(books):
    if book["id"] == "B003":
        remove_index = index
        break
if remove_index is not None:
    removed = books.pop(remove_index)
    print("Removed:", removed)

print(books)
```
