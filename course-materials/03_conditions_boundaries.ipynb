## 1.5.1 Build comparisons that produce Boolean values

A decision begins by evaluating an expression as `True` or `False`. `=` assigns a value; `==` compares for equality. Run `!=`, `<`, `<=`, `>`, and `>=` and inspect their results.

```python
score = 68
print(score >= 50)
print(score == 68)
print(score != 68)
print(type(score >= 50))
```

## 1.5.2 Select between two actions with if and else

A colon ends `if condition:`. Indentation shows which statements belong to the branch. When the condition is False, Python skips the indented block.

```python
score = 68
if score >= 50:
    print("Pass")
print("Decision complete")
```

### else selects one of two actions

`else` handles the case in which the preceding condition is False. The if and else blocks cannot both run in the same decision.

```python
score = 42
if score >= 50:
    result = "Pass"
else:
    result = "Review"
print(result)
```

## 1.5.3 Select among several or independent actions

Python checks conditions from the top and runs only the first True branch. When ranges overlap, put the more selective condition first.

```python
score = 85
if score >= 90:
    grade = "Excellent"
elif score >= 70:
    grade = "Pass"
elif score >= 50:
    grade = "Conditional pass"
else:
    grade = "Review"
print(grade)
```

### Distinguish independent if statements from one chain

Two independent if statements can both run when both conditions are True. An `if / elif / else` chain chooses one candidate. Select the structure that matches the rule.

```python
score = 80
if score >= 50:
    print("Pass")
if score >= 70:
    print("Merit")

if score >= 70:
    label = "Merit"
elif score >= 50:
    label = "Pass"
else:
    label = "Review"
print("Classification:", label)
```

## 1.5.4 Combine decisions with Boolean logic

`and` requires both operands; `or` requires at least one; `not` reverses a Boolean value. For a long expression, use parentheses and meaningful intermediate names, then inspect each part separately.

```python
registered = 35
completion_rate = 72
has_enough_learners = registered >= 30
rate_needs_support = completion_rate < 75
priority = has_enough_learners and rate_needs_support
print(has_enough_learners, rate_needs_support, priority)
print(not priority)
```

## 1.5.5 Validate domains and test boundaries

`>= 50` includes 50; `> 50` does not. For several ranges, test immediately below, exactly at, and immediately above every threshold. Python also supports a chained comparison such as `0 <= score <= 100`.

```python
def support_status(rate):
    if rate < 75:
        return "priority support"
    elif rate < 85:
        return "monitor"
    else:
        return "on track"

for rate in [74.9, 75, 84.9, 85]:
    print(rate, support_status(rate))

score = 100
print(0 <= score <= 100)
```

### Validate a value before classifying it

An invalid value placed in an ordinary category produces a result that runs but is wrong. Check the permitted domain first, then apply the normal classification rules.

```python
score = -5
if not 0 <= score <= 100:
    result = "Invalid score"
elif score >= 50:
    result = "Pass"
else:
    result = "Review"
print(result)
```

## 1.5.6 Use short-circuit evaluation as a guard

`and` does not evaluate its right side when the left side is False; `or` does not evaluate its right side when the left side is True. This can support a safety check, but avoid hiding side effects inside conditions.

```python
registered = 0
completed = 0
if registered > 0 and completed / registered >= 0.75:
    status = "on track"
else:
    status = "needs review"
print(status)
```
