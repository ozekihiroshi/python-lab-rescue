## 1.4.1 Create, inspect, and transform strings

Single and double quotes both create strings. Choose the outer quotes so that quoted content remains readable. `\n` is a newline, `\t` is a tab, and `\\` is a backslash.

```python
course = "Introduction to Python"
message = 'The teacher said, "Run and check."'
print(course)
print(message)
print("Line 1\nLine 2")
```

### Read text as a sequence

String positions start at zero. A slice `[start:end]` excludes the end position. `len()` returns the number of characters.

```python
centre = "North Centre"
print(centre[0])
print(centre[-1])
print(centre[0:5])
print(len(centre))
```

### Clean text without modifying the original object

Strings are immutable. `strip()`, `lower()`, `upper()`, and `replace()` return new strings. Assign the result when it should become the current value.

```python
raw_name = "  North Centre  "
clean_name = raw_name.strip()
print(clean_name)
print(clean_name.lower())
print(clean_name.replace("Centre", "Learning Centre"))
print("Original:", raw_name)
```

## 1.4.2 Combine text and values with compatible types

`+` joins strings and `*` repeats text. Convert a numeric value explicitly with `str()` before concatenation.

```python
completed = 29
print("Completed: " + str(completed))
print("-" * 20)
```

## 1.4.4 Produce meaningful formatted output

Prefix a string with `f` and Python evaluates expressions inside braces. `.1f` displays one decimal place, `.2f` two, and `,` adds a thousands separator.

```python
centre = "North Centre"
registered = 36
completed = 29
rate = completed / registered * 100
cost = 12345.5
print(f"{centre}: {completed}/{registered} learners, completion {rate:.1f}%")
print(f"Materials cost: {cost:,.2f}")
```

## 1.4.5 Read simple delimited text

`split()` separates a string at a delimiter and returns several strings in a list. Collection operations are developed in their own chapter.

```python
record = "North,36,29"
parts = record.split(",")
print(parts)
print(parts[0], int(parts[1]), int(parts[2]))
```
