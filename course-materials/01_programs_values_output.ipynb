## 1.1.1 Read values, expressions, and visible output

Source code contains instructions. The interpreter executes them. An expression evaluates to a value. `print()` makes a requested result visible.

```python
print("Learning centre report")
print(6 * 4)
print(24 - 5)
```

### Identify the parts

- `print(...)` is an **instruction**.
- `6`, `4`, and quoted text are **values**.
- `6 * 4` is an **expression**.
- The three displayed lines are **output**.

A value kind determines possible operations. Lesson 3 teaches type names and conversions systematically.

### Text, expressions, and parentheses

Quoted characters are text. An unquoted numeric expression is calculated. Parentheses are evaluated first.

```python
print("3 + 4")
print(3 + 4)
print((8 + 2) * 5)
```

### Notebook display versus output

A Notebook may display the final expression in a cell. A normal `.py` script does not display an expression by itself. Use `print()` for intended program output.
