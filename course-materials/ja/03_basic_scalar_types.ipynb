## 1.3.1 基本的なスカラー型を区別する

`int`は整数、`float`は小数を含む数、`str`は文字列、`bool`は真偽、`NoneType`は値がない状態を表します。`type()`で実際の型を確認します。

```python
values = [34, 82.5, "34", True, None]
for value in values:
    print(repr(value), type(value))
```

## 1.3.2 数値演算子と明示した順序で計算する

`+`、`-`、`*`、`/`、`//`、`%`、`**`を順に実行します。`/`は通常`float`を返し、`//`は商を床方向へ丸め、`%`は余りを返します。

```python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 2)
```

### 計算順序を明示する

べき乗、乗除算、加減算の順に評価されます。仕事上の意味が読み取れるよう、必要なら括弧を使います。

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
books = 53
per_box = 12
print("満杯の箱:", books // per_box)
print("余る冊数:", books % per_box)
```

## 1.3.3 必要な型へ変換し、値を検証する

`int()`、`float()`、`str()`で変換します。変換できない文字列は`ValueError`になります。`bool()`は内容を読んで真偽を判断するのではなく、空かどうかを基準にするため、`bool("False")`も`True`です。

```python
learners_text = "36"
learners = int(learners_text)
rate_text = "80.5"
rate = float(rate_text)
print(learners + 4, type(learners))
print(rate, type(rate))
print(str(learners) + " learners")
print(bool("False"), bool(""))
```

### 変換エラーを読む

小数を表す文字列`"12.5"`は直接`int()`へ変換できません。まず`float()`へ変換し、その後整数化が本当に必要か判断します。

```python
try:
    int("12.5")
except ValueError as error:
    print(type(error).__name__ + ":", error)

value = float("12.5")
print(value, type(value))
```

### 数値結果を検証する

0で割ると`ZeroDivisionError`になります。浮動小数点数は2進数で近似されるため、`0.1 + 0.2`が画面上で厳密な0.3にならない場合があります。報告表示には`round()`を使えますが、丸める桁と理由を明確にします。

```python
print(0.1 + 0.2)
print(round(0.1 + 0.2, 1))
try:
    print(10 / 0)
except ZeroDivisionError as error:
    print(type(error).__name__ + ":", error)
```

## 1.3.4 統合練習：割合を計算して確かめる

登録者36人、修了者29人です。割る順序、100を掛ける位置、表示する小数桁を予想してから実行します。

```python
registered = 36
completed = 29
completion_rate = completed / registered * 100
print("修了率:", round(completion_rate, 1), "%")
```
