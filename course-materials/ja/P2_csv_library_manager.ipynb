## 最初に仕事の全体像を確認する

1. 配布CSVを読み込む（今回のサンプルは4冊）。

次のセルには原本CSVの実物が表示されます。内容確認用であり、プログラムへ書き写すデータではありません。


```python
from pathlib import Path
import csv

project = Path.cwd() / "projects" / "library-manager"
print('現在の作業フォルダ' + ":", Path.cwd())
print('プロジェクトフォルダ' + ":", project)
print('プロジェクトを確認できた' + ":", project.is_dir())

with (project / "data" / "books.csv").open(encoding="utf-8", newline="") as file:
    for row in csv.DictReader(file):
        print(row)
```
