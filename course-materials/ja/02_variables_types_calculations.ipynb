## 1.2.1 代入によってプログラムの状態を作り、更新する

次のセルを実行する前に、合計と未使用席を予想してください。その後、`morning`の代入だけを20へ変えて再実行します。

```python
capacity = 40
morning = 18
afternoon = 12

print("合計:", morning + afternoon)
print("未使用席:", capacity - (morning + afternoon))
```

### 代入と比較

一つの`=`は右辺を評価して左の名前へ代入します。二つの`==`は値が等しいかを調べ、状態を変更しません。

```python
registered = 40
print("登録者数:", registered)
print("40と等しいか:", registered == 40)
print("35と等しいか:", registered == 35)
```

### 右辺を読んでから左辺を更新する

`total = total + amount`では、右辺の現在値12と5を先に読み、17を作ってから左の`total`を更新します。`+=`は同じ更新の短縮形です。

```python
total = 12
amount = 5
print("更新前:", total)
total = total + amount
print("長い形の後:", total)
total += amount
print("短い形の後:", total)
```

### 計算済みの値は自動更新されない

`delivered`を計算した後で`cancelled`だけを変えても、`delivered`は自動では変わりません。二つの表示を予想してから実行してください。

```python
planned = 15
cancelled = 2
delivered = planned - cancelled

cancelled = 4
print("再計算前:", delivered)

delivered = planned - cancelled
print("再計算後:", delivered)
```
