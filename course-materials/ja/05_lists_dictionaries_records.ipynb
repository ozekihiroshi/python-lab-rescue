## 2.1.1 リストで順序のある値を保持し、変更する

角括弧で作り、0から始まる添字、負の添字、スライスで値を読みます。実務では同じ意味の値をまとめると処理しやすくなります。

```python
attendance = [28, 31, 30, 33]
print(attendance[0])
print(attendance[-1])
print(attendance[1:3])
print(len(attendance))
print(31 in attendance)
```

### リストの内容を変更する

添字への代入、`append()`、`extend()`、`insert()`、`remove()`、`pop()`で内容を変えられます。変更メソッドの多くはリスト自身を変更し、便利な新リストを返すわけではありません。

```python
attendance = [28, 31, 30]
attendance[0] = 29
attendance.append(33)
attendance.extend([32, 34])
removed = attendance.pop()
print(attendance)
print("取り出した値:", removed)
```

### 代入は同じリストを共有し、copy()は別のリストを作る

`backup = original`は二つの名前が同じリストを指します。一方を変更すると他方からも見えます。独立した浅いコピーには`copy()`を使います。

```python
original = [28, 31, 30]
shared = original
separate = original.copy()
shared.append(33)
print("original:", original)
print("shared:", shared)
print("separate:", separate)
```

### 位置が不要なら値を直接、必要ならenumerate()で読む

リストを処理するために添字を手作業で管理する必要はありません。値だけならfor、週番号も必要ならenumerateを使います。

```python
attendance = [28, 31, 30, 33]
for week, value in enumerate(attendance, start=1):
    print(f"第{week}週: {value}人")
```

## 2.1.2 タプルで固定された順序付き値を表す

丸括弧で作り、添字やループで読めます。座標、固定された月と年度など、要素の組を途中で変更させたくない場合に使います。

```python
report_period = (2026, 8)
year, month = report_period
print(year, month)
# report_period[0] = 2027  # TypeError: タプルは変更不能
```

## 2.1.3 辞書で名前付きフィールドを表す

辞書は`キー: 値`の組です。添字の位置ではなく、`centre["registered"]`のような名前で値を読みます。同じ辞書内のキーは重複できません。

```python
centre = {
    "name": "North Learning Centre",
    "district": "North",
    "registered": 40,
    "completed": 32,
}
print(centre["name"])
print(centre["completed"] / centre["registered"] * 100)
```

### 存在しないキーをどのように扱うか決める

角括弧で存在しないキーを読むと`KeyError`です。任意項目なら`get()`で既定値を指定できます。必須項目にgetを使って欠落を隠さず、業務上の必須・任意を区別します。

```python
centre = {"name": "North", "registered": 40}
print(centre.get("phone", "未登録"))
print("completed" in centre)
centre["completed"] = 32
centre["registered"] = 42
print(centre)
```

### items()でキーと値を同時に反復する

辞書を直接for文へ渡すとキーを反復します。キーと値の両方が必要なら`items()`、キーだけなら`keys()`、値だけなら`values()`を使います。

```python
centre = {"registered": 40, "attended": 34, "completed": 32}
for field, value in centre.items():
    print(field, value)
```

## 2.1.4 辞書のリストで複数レコードを表す

一つの辞書を一件のレコード、一つのキーを項目名として扱えます。これは後でCSVを表やDataFrameとして扱うための橋になります。

```python
centres = [
    {"name": "North", "district": "A", "registered": 40, "completed": 32},
    {"name": "South", "district": "B", "registered": 35, "completed": 24},
]
for centre in centres:
    rate = centre["completed"] / centre["registered"] * 100
    print(f"{centre['name']}: {rate:.1f}%")
```

## 2.1.5 集合で重複のない所属を表す

集合は順序や重複ではなく、ある値が含まれるかを扱います。和集合`|`、積集合`&`、差集合`-`でカテゴリの関係を比較できます。表示順には依存しません。

```python
offered = {"Python", "Data", "Office"}
requested = {"Python", "Web", "Data"}
print("共通:", offered & requested)
print("未提供:", requested - offered)
print("すべて:", offered | requested)
```

### 構造を目的で選ぶ

順序付きで変更する値はリスト、固定された組はタプル、名前付き項目は辞書、重複のない所属は集合です。複数のレコードはリストの中に辞書を置いて表せます。

## 2.1.6 備品台帳のレコードを検索・追加・更新・削除する

`asset_id`、`name`、`available`を持つ三件の備品を辞書のリストで表してください。IDで一件を探し、見つからない場合は`None`とします。重複しない`A004`を追加し、`A002`を利用可能へ変更し、`A003`を削除します。最後にID順と件数を表示し、元の三件だけを保持した浅いコピーとの違いも確認してください。


```python
books = [
    {"id": "B001", "title": "Python Basics", "read": False},
    {"id": "B002", "title": "Working with Data", "read": True},
    {"id": "B003", "title": "Useful Functions", "read": False},
]

target_id = "B002"
found = None
for book in books:
    if book["id"] == target_id:
        found = book
        break

print(found)
```

### 同じレコード集合を追加・参照・更新・削除する

追加前にはID集合で重複を確認します。検索で返された辞書はリスト内の辞書そのものなので、その項目を変更すると保存されたレコードも変わります。削除では`enumerate()`で位置とレコードを同時に受け取り、その位置を`pop()`します。先頭から探すことで、残るレコードの順序を壊しません。これらをCreate、Read、Update、Deleteの頭文字からCRUDと呼びます。

```python
existing_ids = set()
for book in books:
    existing_ids.add(book["id"])

new_book = {"id": "B004", "title": "Files in Practice", "read": False}
if new_book["id"] not in existing_ids:
    books.append(new_book)

found["read"] = False

remove_index = None
for index, book in enumerate(books):
    if book["id"] == "B003":
        remove_index = index
        break
if remove_index is not None:
    removed = books.pop(remove_index)
    print("削除:", removed)

print(books)
```
