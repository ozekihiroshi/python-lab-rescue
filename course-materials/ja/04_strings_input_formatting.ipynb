## 1.4.1 文字列を作り、調べ、変換する

一重引用符と二重引用符はどちらも文字列を作ります。内容に引用符を含める場合は外側と使い分けます。`\n`は改行、`\t`はタブ、`\\`はバックスラッシュです。

```python
course = "Python入門"
message = '講師は「実行して確認」と言いました。'
print(course)
print(message)
print("1行目\n2行目")
```

### 文字の並びとして読む

文字列は0から始まる位置で文字を取り出せます。スライス`[開始:終了]`は終了位置を含みません。`len()`は文字数を返します。

```python
centre = "North Centre"
print(centre[0])
print(centre[-1])
print(centre[0:5])
print(len(centre))
```

### 元の文字列を壊さずに整える

文字列は変更不能です。`strip()`、`lower()`、`upper()`、`replace()`は整えた新しい文字列を返します。必要なら名前へ再代入します。

```python
raw_name = "  North Centre  "
clean_name = raw_name.strip()
print(clean_name)
print(clean_name.lower())
print(clean_name.replace("Centre", "Learning Centre"))
print("元の値:", raw_name)
```

## 1.4.2 型を合わせて文字列と値を組み合わせる

`+`は文字列同士を連結し、`*`は文字列を繰り返します。数値を連結する場合は`str()`で明示的に変換します。

```python
completed = 29
print("修了者: " + str(completed))
print("-" * 20)
```

## 1.4.3 入力を受け取り、計算用の型へ変換する

`input()`は、数字を入力しても常に文字列を返します。下のセルは自動検証で停止しないようコメントにしてあります。新しいセルへコピーしてコメントを外し、`36`を入力して型を確認してください。

## 1.4.4 意味の分かる書式付き出力を作る

文字列の先頭に`f`を付けると、波括弧内の式を評価できます。`.1f`は小数第1位、`.2f`は小数第2位、`,`は桁区切りです。

```python
centre = "North Centre"
registered = 36
completed = 29
rate = completed / registered * 100
cost = 12345.5
print(f"{centre}: {completed}/{registered}人、修了率 {rate:.1f}%")
print(f"教材費: {cost:,.2f}")
```

## 1.4.5 単純な区切り文字付きテキストを読む

`split()`は区切り文字で文字列を分け、複数の文字列を返します。戻り値はリストなので、詳しい操作はコレクションの章で学びます。

```python
record = "North,36,29"
parts = record.split(",")
print(parts)
print(parts[0], int(parts[1]), int(parts[2]))
```
