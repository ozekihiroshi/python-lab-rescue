在庫CSVと当日の依頼CSVを確認してから、用意済みのスターターを完成させます。提出対象は `equipment_lending.py` だけです。

```python
from pathlib import Path
import csv

project = Path.cwd() / "projects" / "equipment-lending"
inventory = project / "data" / "equipment_inventory.csv"
requests = project / "data" / "lending_requests.csv"
print("Project:", project)
print("Inventory found:", inventory.is_file())
print("Requests found:", requests.is_file())
```

## 1. 原本を確認する

下のセルは、プログラムへ直接書き写す答えではありません。開始時の状態と、順番に処理すべき仕事を確認するために表示します。

```python
for label, path in [("OPENING INVENTORY", inventory), ("TODAY'S REQUESTS", requests)]:
    print("\n" + label)
    with path.open(newline="", encoding="utf-8") as handle:
        for row in csv.DictReader(handle):
            print(row)
```
