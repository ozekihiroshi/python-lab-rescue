## 1.1.1 値・式・出力を読み分ける

ソースコードには命令があります。インタプリタが命令を実行し、式は一つの値へ評価され、`print()`が結果を外へ示します。

```python
print("学習センター報告")
print(6 * 4)
print(24 - 5)
```

### 各部分を見分ける

- `print(...)`は**命令**です。
- `6`、`4`、引用符内の文字は**値**です。
- `6 * 4`は**式**です。
- 表示された三行は**出力**です。

値の種類は可能な操作を決めます。型名と変換はレッスン3で体系的に学びます。

### 文字・式・括弧

引用符内は文字です。引用符のない数値式は計算されます。括弧内は先に評価されます。

```python
print("3 + 4")
print(3 + 4)
print((8 + 2) * 5)
```

### Notebookの表示と出力

Notebookはセル末尾の式を補助表示することがあります。通常の`.py`スクリプトは式だけでは表示しません。意図した出力には`print()`を使います。
