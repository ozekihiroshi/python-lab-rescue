## 1.6.1 forとrangeで既知の値を繰り返し処理する

`for 一時的な名前 in 値の並び:` と書き、字下げされた処理を各要素について一回ずつ実行します。ループ変数は現在処理している値を表します。

```python
weekly_attendance = [28, 31, 32, 34]
for attendance in weekly_attendance:
    print("今週の出席者数:", attendance)
print("処理完了")
```

### range()で回数と整数の範囲を表す

`range(start, stop, step)` はstartを含み、stopを含みません。`range(1, 5)`は1、2、3、4です。反復回数を確認すると、境界の一つ多い・少ないという誤りを防げます。

```python
print(list(range(1, 5)))
for week in range(1, 5):
    print(f"第{week}週")
print(list(range(10, 4, -2)))
```

## 1.6.2 合計と条件付き件数を作る

合計のように反復をまたいで残す値をアキュムレータと呼びます。最初は0にし、各反復で現在の値を加えます。途中経過を表示すると更新順序を追跡できます。

```python
weekly_attendance = [28, 31, 32, 34]
total = 0
for attendance in weekly_attendance:
    total += attendance
    print("途中の合計:", total)
print("合計:", total)
print("平均:", total / len(weekly_attendance))
```

### 条件に一致した件数を数える

カウンタも0から始め、条件がTrueのときだけ1増やします。合計する値と数える件数を別の変数にすると、役割が明確になります。

```python
weekly_attendance = [28, 31, 32, 34]
weeks_at_least_30 = 0
for attendance in weekly_attendance:
    if attendance >= 30:
        weeks_at_least_30 += 1
print(weeks_at_least_30)
```

## 1.6.3 最大・最小などの値と位置を追跡する

最大値を0で初期化すると、すべて負のデータで誤ります。空でないことが分かっているなら最初の要素から始めます。空データを許す場合は、先に明示的に扱います。

```python
weekly_change = [-4, -2, -7, -3]
if weekly_change:
    largest = weekly_change[0]
    for change in weekly_change[1:]:
        if change > largest:
            largest = change
    print("最大値:", largest)
else:
    print("データなし")
```

### enumerate()で位置と値を同時に得る

番号と値の両方が必要なら、手作業のカウンタより`enumerate()`を使います。`start=1`を指定すると、人が読む第1週から始められます。

```python
weekly_attendance = [28, 31, 32, 34]
for week, attendance in enumerate(weekly_attendance, start=1):
    print(f"第{week}週: {attendance}人")
```

## 1.6.4 一回の処理またはループ全体を制御する

`continue`は現在の反復の残りを飛ばして次へ進みます。`break`はループ全体を終了します。通常処理と例外的な制御を分け、なぜ飛ばす・止めるのかが読めるようにします。

```python
readings = [28, None, 31, -1, 34]
for reading in readings:
    if reading is None:
        continue
    if reading < 0:
        print("無効値を検出。処理を停止します")
        break
    print("有効値:", reading)
```

## 1.6.5 条件が変わるまでwhileで繰り返す

処理回数より終了条件で反復を決める場合は`while`を使います。条件に使う状態をループ内で更新しないと、無限ループになる可能性があります。

```python
remaining = 3
while remaining > 0:
    print("残り:", remaining)
    remaining -= 1
print("完了")
```

## 1.6.6 0回の繰り返しと空の入力を扱う

処理する値の並びがあるならfor、条件が満たされるまで続けるならwhileが自然です。既知のデータを添字でwhile処理するより、値を直接読むforの方が境界エラーを減らせます。

```python
weekly_attendance = [28, 31, 32, 34]
for attendance in weekly_attendance:
    print(attendance)

# 同じデータを添字で扱うwhileは、更新と境界の管理が必要です。
index = 0
while index < len(weekly_attendance):
    print(weekly_attendance[index])
    index += 1
```
