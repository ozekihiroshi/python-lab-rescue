## What a program does

A computer does not infer intention. It follows explicit instructions.

**Input → processing in memory → output and/or persistent storage**

The Python interpreter reads source code and executes it. Valid syntax, successful execution, and a result that matches the purpose are three different questions.

## Three execution methods

- **Notebook (course standard):** saves explanation, code, and visible results.
- **Console:** runs one entered statement immediately.
- **`.py` script:** normally runs saved instructions from top to bottom.

Values, variables, decisions, loops, and functions transfer to other languages. Translation, type rules, and memory management differ.
